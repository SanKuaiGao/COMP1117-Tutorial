## Task 1: Factorial

**Core** - about 7 minutes

Return n factorial as an integer. Use a recursive call to factorial. The product of no factors is 1, so factorial(0) must return 1.

**Input:** n is an integer with 0 <= n <= 20. Input validation is not required.

**Implementation:** Use recursion. Do not use loops, math.factorial, or a table of precomputed answers.

Return the required value. Do not print inside your function and do not use global mutable state.

### Examples

- `factorial(0)` gives `1`.
- `factorial(1)` gives `1`.
- `factorial(5)` gives `120`.
- `factorial(8)` gives `40320`.

### Before coding

1. What should factorial(0) return, and why?
2. What work is waiting when factorial(3) calls factorial(2)?
3. What changes if the last line calls factorial but does not return its result?

Write down a base case, a smaller input, and how the returned result completes the current call.

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

### Explanation

The base case returns the empty product, 1. For n > 0, the remaining product is factorial(n - 1). Multiplying that returned value by n completes the current call. n decreases by one until it reaches zero.

**Common mistakes:** Returning 0 at the base case makes every result zero. Omitting return loses the computed result. Calling factorial(n) again does not reduce the problem.

```python
assert factorial(0) == 1
assert factorial(1) == 1
assert factorial(5) == 120
assert factorial(8) == 40320
print("Task 1: example checks passed")
```

## Task 2: Sum of digits

**Core** - about 8 minutes

Return the sum of the decimal digits of a non-negative integer. Use integer division and remainder to separate the last digit from the remaining digits.

**Input:** n is an integer with 0 <= n <= 10**12. Input validation is not required.

**Implementation:** Use recursion. Do not convert n to a string, use loops, or use sum().

Return the required value. Do not print inside your function and do not use global mutable state.

### Examples

- `sum_digits(0)` gives `0`.
- `sum_digits(7)` gives `7`.
- `sum_digits(2026)` gives `10`.
- `sum_digits(1002003)` gives `6`.

### Before coding

1. What are 2026 // 10 and 2026 % 10?
2. Which call handles zero digits inside 1002?
3. Give an input with repeated digits and predict its answer before running it.

Write down a base case, a smaller input, and how the returned result completes the current call.

```python
def sum_digits(n):
    if n < 10:
        return n
    return sum_digits(n // 10) + n % 10
```

### Explanation

A one-digit number is already its digit sum. For a larger number, n % 10 is the last digit and n // 10 removes it. The recursive call sums the remaining digits. Adding the last digit gives the required result.

**Common mistakes:** Use //, not /, so the smaller input remains an integer. n - 1 removes one unit rather than one decimal digit. The input zero needs a defined result.

```python
assert sum_digits(0) == 0
assert sum_digits(7) == 7
assert sum_digits(2026) == 10
assert sum_digits(1002003) == 6
print("Task 2: example checks passed")
```

## Task 3: Choosing a group

**Core** - about 10 minutes

Return the number of ways to choose k distinct items from n distinct items when order does not matter. Return 0 when k > n. Choosing no items, or choosing all n items, has one possible result.

**Input:** n and k are integers with 0 <= n <= 12 and 0 <= k <= 14. Input validation is not required.

**Implementation:** Use combinations(n - 1, k - 1) + combinations(n - 1, k) for the recursive case. Do not use loops, factorial, math.comb, or memoization.

Return the required value. Do not print inside your function and do not use global mutable state.

### Examples

- `combinations(0, 0)` gives `1`.
- `combinations(5, 2)` gives `10`.
- `combinations(6, 3)` gives `20`.
- `combinations(2, 5)` gives `0`.

### Before coding

1. Explain the two recursive calls in terms of including or excluding one item.
2. Why does combinations(0, 0) return 1?
3. Which subproblem appears twice while computing combinations(4, 2)?

Write down a base case, a smaller input, and how the returned result completes the current call.

```python
def combinations(n, k):
    if k > n:
        return 0
    if k == 0 or k == n:
        return 1
    return (combinations(n - 1, k - 1)
            + combinations(n - 1, k))
```

### Explanation

Pick one distinguished item. A group either includes it, leaving k - 1 choices from n - 1 items, or excludes it, leaving k choices from n - 1 items. The two cases are disjoint, so their counts add. Handle impossible and boundary choices before recursing.

**Common mistakes:** Both k = 0 and k = n need base cases. k > n must return 0. Multiplying the two recursive results counts a different problem. Repeated subproblems make large inputs slow.

```python
assert combinations(0, 0) == 1
assert combinations(5, 2) == 10
assert combinations(6, 3) == 20
assert combinations(2, 5) == 0
print("Task 3: example checks passed")
```

## Task 4: Writing a number in another base

**Core** - about 12 minutes

Return the representation of n in base b as a string. Use digits 0-9 and uppercase A-F. Return '0' when n is zero. Positive numbers must have no leading zeros.

**Input:** n and b are integers with 0 <= n <= 10**12 and 2 <= b <= 16. Input validation is not required.

**Implementation:** Use recursion, //, %, and the string '0123456789ABCDEF'. Do not use loops, bin(), oct(), hex(), or formatting functions for base conversion.

Return the required value. Do not print inside your function and do not use global mutable state.

### Examples

- `convert_base(0, 2)` gives `'0'`.
- `convert_base(35, 2)` gives `'100011'`.
- `convert_base(255, 16)` gives `'FF'`.
- `convert_base(47, 3)` gives `'1202'`.

### Before coding

1. For n = 35 and b = 2, what are the quotient and remainder?
2. Why is the remainder appended after the recursive result?
3. Which tests distinguish '0', an empty string, and an unwanted leading zero?

Write down a base case, a smaller input, and how the returned result completes the current call.

```python
def convert_base(n, b):
    digits = "0123456789ABCDEF"
    if n < b:
        return digits[n]
    return convert_base(n // b, b) + digits[n % b]
```

### Explanation

When n < b, one digit is enough, including the digit for zero. Otherwise the quotient n // b contains the earlier digits and the remainder n % b gives the last digit. Append the last digit after the recursive result to preserve the order.

**Common mistakes:** A base case that returns an empty string for zero does not meet this task's zero-input contract. Prepending each remainder reverses the digits. str(remainder) does not produce A-F.

```python
assert convert_base(0, 2) == '0'
assert convert_base(35, 2) == '100011'
assert convert_base(255, 16) == 'FF'
assert convert_base(47, 3) == '1202'
print("Task 4: example checks passed")
```

## Task 5: Alternating digit weights

**Optional extension**

Write two mutually recursive functions, sum_odd(n) and sum_even(n). Count digit positions from the right, starting at 1. sum_odd doubles digits in odd positions. sum_even doubles digits in even positions. Digits in the other positions keep their value. Return the resulting sum.

**Input:** n is an integer with 0 <= n <= 10**12. Input validation is not required.

**Implementation:** Each function must call the other for the remaining digits. Use // and %. Do not convert to a string or use a loop. Odd/even refers to the position, not the digit value. Add doubled digits directly, even when a doubled digit exceeds 9.

Return the required value. Do not print inside your function and do not use global mutable state.

### Examples

- `sum_odd(0), sum_even(0)` gives `(0, 0)`.
- `sum_odd(7), sum_even(7)` gives `(14, 7)`.
- `sum_odd(12), sum_even(12)` gives `(5, 4)`.
- `sum_odd(1234), sum_even(1234)` gives `(16, 14)`.

### Before coding

1. Why must the next recursive call switch function names?
2. What should the pair of functions return for the single digit 9?
3. Trace sum_odd(1234), recording which function receives each smaller input.

Write down a base case, a smaller input, and how the returned result completes the current call.

```python
def sum_odd(n):
    if n < 10:
        return 2 * n
    return 2 * (n % 10) + sum_even(n // 10)

def sum_even(n):
    if n < 10:
        return n
    return n % 10 + sum_odd(n // 10)
```

### Explanation

Removing the rightmost digit changes the parity of every remaining position relative to the new right edge. The next call therefore switches functions. For 1234, sum_odd is 2*4 + 3 + 2*2 + 1 = 16.

**Common mistakes:** Doubling odd digit values solves a different problem. Calling the same function keeps the weight unchanged. Do not split a doubled digit into decimal digits for this exercise.

```python
assert sum_odd(0) == 0
assert sum_even(0) == 0
assert sum_odd(7) == 14
assert sum_even(7) == 7
assert sum_odd(12) == 5
assert sum_even(12) == 4
assert sum_odd(1234) == 16
assert sum_even(1234) == 14
print("Task 5: example checks passed")
```

## Task 6: Nested recursive calls

**Optional extension**

Implement these rules: A(0, n) = n + 1. For m > 0, A(m, 0) = A(m - 1, 1). For m > 0 and n > 0, A(m, n) = A(m - 1, A(m, n - 1)).

**Input:** For this exercise and all provided tests, m and n are integers with 0 <= m <= 2 and 0 <= n <= 6. Input validation is not required. Keep your own examples within this range.

**Implementation:** Translate the recurrence using recursive calls. Do not replace it with a closed-form formula or use loops.

Return the required value. Do not print inside your function and do not use global mutable state.

### Examples

- `ackermann(0, 0)` gives `1`.
- `ackermann(0, 4)` gives `5`.
- `ackermann(1, 2)` gives `4`.
- `ackermann(2, 3)` gives `9`.

### Before coding

1. Which rule applies to A(0, 0)?
2. What is evaluated first in A(0, A(1, 0))?
3. Trace A(1, 1) without running the code.

Write down a base case, a smaller input, and how the returned result completes the current call.

```python
def ackermann(m, n):
    if m == 0:
        return n + 1
    if n == 0:
        return ackermann(m - 1, 1)
    return ackermann(m - 1, ackermann(m, n - 1))
```

### Explanation

The first case must be checked before the n = 0 case. In the final branch, evaluate the inner call A(m, n - 1) first, then use its result as the second argument of A(m - 1, ...). This example illustrates nested recursive evaluation.

**Common mistakes:** The outer call needs the returned value of the inner call. Swapping arguments changes the function. Values outside the stated exercise range can require much more work, so the reference notebook's large-input demonstration is not included.

```python
assert ackermann(0, 0) == 1
assert ackermann(0, 4) == 5
assert ackermann(1, 2) == 4
assert ackermann(2, 3) == 9
print("Task 6: example checks passed")
```
